```python
# Encapsulation
# Encapsulation is the bundling of data and methods that operate on that data within a single unit,
#  such as a class in object-oriented programming. It restricts direct access to some of an object's components, 
# which can prevent the accidental modification of data and promote modularity.
```

```python
## Encapsulation with Getter and Setter methods 
# Public, private and protected attributes

class Person:
    def __init__(self, name, age):
        self.name = name # public attribute
        self.age = age # public attribute

person = Person('Sau', 24)

print(person.name)
print(person.age)

dir(person)
        
```

```python
## Encapsulation with Private attributes 
class Person:
    def __init__(self, name, age):
        self.__name = name # private attribute
        self.__age = age # private attribute

person = Person('Sau', 24)

dir(person)
# print(person.__name) # AttributeError: 'Person' object has no attribute '__name'
```

```python
class Person:
    def __init__(self, name, age):
        self._name = name # protected attribute
        self._age = age # protected attribute

person = Person('Sau', 24)
print(person._name)   # protected — convention(should be accessed from derived class like subclass ), it can be accessed from outside 
```

```python
# Encapsulation with getter and setter 

class Person:
    def __init__(self, name, age):
        self.__name = name # private attribute
        self.__age = age # private attribute
        
    def get_name(self):
        return self.__name
    
    def set_name(self, name):
        self.__name = name

    def get_age(self):
        return self.__age

    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age must be positive.")

person =  Person("Sau", 24)

## Access and modify private variables using getter and setter

print(person.get_age())
print(person.get_name())

person.set_age(23)
print(person.get_age())

person.set_name("Harshi")
print(person.get_name())

person.set_age(-5)
```
