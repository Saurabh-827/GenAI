```python
# Abstraction 
# It is a process of hiding the implementation details and showing only functionality to the user.
# It is achieved by using abstract classes and abstract methods.
# Interfaces are not present in python
```

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    def drive(self):
        return "This vehicle used for driving"
    
    # abstract method — must be implemented by subclasses
    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        return "Car engine started"
    
def operate_vehicle(vehicle):
    return vehicle.start_engine(), vehicle.drive()

car = Car()

print(operate_vehicle(car))
```
