```python
### Instance variable and methods 
class dog:
    ## constructor
    def __init__(own, name,age):   # i didn't use self(object) to remember that it can be anything
        own.name=name
        own.age=age

dog1 = dog("tommy",2)
print(dog1.name)
print(dog1.age)
```

```python
# define a class with instance methods 

class student:
    def __init__(self, name, age):
        self.name=name
        self.age=age
    def show(self):
        print("name:",self.name)
        print("age:",self.age)

s1 = student("John", 50)
print(s1)
s1.show()
```

```python
## Modeling a bank account

class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount
        print(f'Updated balance is {self.balance}')
    
    def withdraw(self, amount):
        if amount > self.balance:
            print('Insufficient Funds')
        else:
            self.balance -= amount
            print(f'updated balance is {self.balance}')

    def getBalance(self):
        return self.balance
    
account1 = BankAccount("Ice", 1000)

account1.getBalance()

account1.deposit(500)
account1.withdraw(2000)
account1.withdraw(300)
```
