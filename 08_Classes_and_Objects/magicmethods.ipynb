```python
# Magic Methods in Python are special methods that allow you to define the behavior of your objects in various contexts. 
# They are also known as "dunder" methods (double underscore) because they are surrounded by double underscores.
# These methods enable you to implement custom behavior for built-in operations such as addition, subtraction, string representation, and more.
```

```python
#Basic Methods 

class Person:
    def __init__(self, name, age ):
        self.name = name
        self.age = age

person = Person("Sau", 23)
print(person.name)
        
```

```python
class Person:
    def __init__(self, name, age ):
        self.name = name
        self.age = age

# __str__ — called when print() is used, human readable format
    def __str__(self):
        return f"{self.name} is {self.age} years old"
# __repr__ — called when repr() is used, developer/debugging format
    def __repr__(self):
        return f"Person(name={self.name}, age={self.age})"

person = Person("Sau", 23)
print(person)
print(repr(person))
```
