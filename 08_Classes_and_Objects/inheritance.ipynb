```python
## Inheritance in python
# Inheritance is a fundamental concept in object-oriented programming (OOP) that allows 
# a new class (called a child or subclass) to inherit attributes and behaviors (methods) from an existing class (called a parent or superclass).
# This promotes code reusability and establishes a natural hierarchical relationship between classes.
```

```python
# creating a parent class 

class Animal:
    def __init__(self,name, species):
        self.name = name
        self.species = species
    def speak(self):
        return "Animal speaks"

Animal1 = Animal("Dog", "Canine")
print(Animal1.name)  
print(Animal1.species)
print(Animal1.speak())
```

```python
# creating a child class
# here single inheritance
class Dog(Animal):
    def __init__(self, name, species, breed):
        super().__init__(name,species)
        self.breed = breed
    def speak(self):
        print("Woof!") # does not return a value, so calling it inside print() prints the output followed by None   
    
Dog1 = Dog("Buddy", "Canine", "Golden Retriever")
print(Dog1.name)
print(Dog1.species)
print(Dog1.breed)
print(Dog1.speak()) # This will call the overridden speak method in the Dog class
```

```python
## multiple inheritance

# base class 2

class CanFly:
    def __init__(self,flyingAbility):
        self.flyingAbility = flyingAbility

# derived class from both Animal and CanFly 

class Bird(Animal, CanFly):
    def __init__(self, name, species, flyingAbility):
        Animal.__init__(self,name, species)
        CanFly.__init__(self, flyingAbility)    # better practice super().__init__(name, species, flyingAbility)
    def speak(self):
        return "Chirp!"
    
Bird1 = Bird("Sky", "Avian", "Yes")
print(Bird1.name, Bird1.species, Bird1.flyingAbility, Bird1.speak())
```
