```python
## Loops in Python 

## For loop

range(5)
```

```python
#For Loop 

for i in range(5):
    print(i)
```

```python
for i in range(1,8):
    print(i)
```

```python
## increment and decrement in python
#default increment is 1

for i in range(1, 10 , 3):    #incremented by 3 
    print(i)


for i in range(20 ,  8, -4):  #incremented by -4 (overall decremented)
    print(i)
```

```python
## Strings

 # In Python, we can traverse a string using a loop to iterate over each character
str = "Saurabh Harshita"

for i in str:
    print(i)
```

```python
## While loop

num = 1

while num <= 5:
    print(num)
    num += 1
```

```python
## Loop Control Statements


# Break statement
 # Break statement helps to exit the loop before it completes 

for i in range(20):
    if i == 10:
        break
    print(i)

```

```python
## continue 

# Continue statement is used to skip the current iteration of the loop and move to the next iteration

# Example: Here we print odd numbers and use continue to skip even numbers
for i in range(20):
    if i % 2 == 0:
        continue
    print(i)
```

```python
## Pass

# Pass statement is a null statement in Python. It is used when a statement is required syntactically but you do not want any command or code to execute.

for i in range(5):   # it is written here but does nothing
    pass
print("This will execute");
```

```python
## Nested loops 
# loop inside loop

for i in range(4):
    for j in range(3):
        print(f"i:{i} and j:{j}")

```

```python
## Example: Calculate the sum of first N natural numbers using a while and for loop

# using while loop

num = int(input("Enter a number: "))
sum = 0
count = 1
while(count <= num):
    sum += count
    count += 1
print("Total sum:", sum)

# using for loop
sum = 0
for i in range(1, num + 1):
    sum += i
print("Total sum:", sum)
```

```python
## Example- Prime numbers between 1 and 100

for num in range(1, 101):
    if num > 1:
        for i in range(2, num):
            if num % i == 0:
                break
        else:                 #here we use else with the for loop
            print(num)
```
