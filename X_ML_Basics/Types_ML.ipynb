```python
## Types of Machine learning
# Supervised learning -  Regression, Classification
# Unsupervised learning -  Clustering, Anomaly Detection, Dimensionality Reduction, Association Rule Learning
# Reinforcement learning
# Semi-supervised learning

## Supervised Learning
# In supervised learning, the training data is provided with the correct answers, 
# allowing the model to learn the relationship between input features and target variables.
# It can be further categorized into:
# 1. Regression: Predicts continuous values (e.g., Next day market prices).
# 2. Classification: Predicts discrete categories (categories are defined during traing) (e.g., spam detection).

## Unsupervised Learning
# Unsupervised learning deals with unlabeled data, where the model tries to find hidden patterns or structures.
# Common tasks include:
# 1. Clustering: Groups similar data points together (it makes clusters of similar data) (e.g., customer segmentation).
# 2. Anomaly Detection: Identifies unusual data points (e.g., fraud detection).
# 3. Dimensionality Reduction: Reduces the number of features while preserving important information (e.g., PCA).
# 4. Association Rule Learning: Discovers interesting relationships in large datasets (e.g., market basket analysis).

## Reinforcement Learning
# Reinforcement learning involves an agent that learns to make decisions by performing actions in an environment to maximize a reward.
# It is commonly used in robotics, gaming, and autonomous systems.

## Semi-Supervised Learning
# Semi-supervised learning uses a combination of labeled and unlabeled data for training. 
# It is useful when obtaining labeled data is expensive or time-consuming.
```
