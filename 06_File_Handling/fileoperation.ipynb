```python
# File Operations:  Read and write files 

#  Read a file 

with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
## Read file line by line

with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip())  # strip removes new line character 
```

```python
# writing a file (overwrite)

with open('example.txt','w') as file:
    file.write("Hello this is a write operation\n")
    file.write("this is last line added")
```

```python
# write a file (without overwrite)

with open('example.txt', 'a') as file:
    file.write("This is an append line\n")
```

```python
# writing a list of lines 

lines = ['\nLine1\n', 'Line2\n', 'Line3']

with open('example.txt', 'a') as file:
    file.writelines(lines)
```

```python
# binary files operation
data = b'\x00\x01\x02\x03\x04'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
# read files operation

with open('example.bin', 'rb') as file:
    content = file.read()
print(content)
```

```python
# read from source file and write it in destination file 

with open('example.txt', 'r') as file:
    content = file.read()
with open('destination.txt', 'w') as file:
    file.write(content)
```

```python
# count lines words and characters in a file 

def count_everything(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(word.split()) for word in lines)
        char_count = sum(len(word) for word in lines)
    return line_count, word_count, char_count

lines, words , chars = count_everything('example.txt')
print(f"Lines: {lines}, Words: {words}, Characters: {chars}")
```

```python
## writing and again reading file

with open('example.txt', 'w+') as file:
    file.write("This is first line\n")
    file.write("This is second line\n")

    # move the cursor at beginning
    file.seek(0)
    #read file 
    content =  file.read()
    print(content)
```
