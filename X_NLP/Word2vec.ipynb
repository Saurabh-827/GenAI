```python
## Word Embeddings
# Word Embeddings are a type of word representation that allows words to be represented as vectors in a continuous vector space.
# The idea is to capture the semantic meaning of words in a way that similar words are represented by similar vectors.
# For example, the words "king" and "queen" might be represented by vectors that are close to each other in the vector space, while the word "cat" might be represented by a vector that is farther away.
# Word embeddings are typically learned from large corpora of text using unsupervised learning techniques.
## What is word2vec?
# word2vec is a popular algorithm for learning word embeddings.

# Advantages of word2vec
# 1. low dimensionality (like 100 to 300 ) -It is computationally efficient and can be trained on large corpora of text.
# 2. It captures semantic relationships between words, such as analogies (e.g.,
# 3. dense vector (handle non-zero)
```

```python
import gensim 
from gensim.models import Word2Vec, KeyedVectors

# 1. Humara Corpus (Text Data)
# Word2Vec ko data list of lists ke form mein chahiye hota hai (har sentence ek list)
sentences = [
    ['i', 'love', 'machine', 'learning'],
    ['deep', 'learning', 'is', 'complex', 'but', 'good'],
    ['machine', 'learning', 'is', 'the', 'future'],
    ['i', 'love', 'coding', 'and', 'learning'],
    ['coding', 'in', 'python', 'is', 'fun']
]

# 2. Model Train Karna
# min_count=1 : Un words ko bhi lo jo kam se kam 1 baar aaye hain
# vector_size=5 : Har word ko 5 numbers (dimensions) ke array mein convert karo
# window=2 : Ek word ke aage aur peeche ke 2 words ko context maano

model = Word2Vec(sentences, min_count=1, vector_size=5, window=2)

# 3. Model se sawaal poochna (The Magic!)

# Dekhte hain 'learning' word ka actual math/vector kaisa dikhta hai

print("Vector for 'learning':")
print(model.wv['learning'])
print("\n" + "-"*50 + "\n")

# Dekhte hain 'learning' ke sabse similar (milte-julte) words kaunse hain
print("Most similar words to 'learning':")
similar_words = model.wv.most_similar('learning')
for word , similarity in similar_words:
    print(f"{word} : {similarity:.4f}")
```

```python
## types of Word2vec

## CBOW
# CBOW (Continuous Bag of Words) ek architecture hai jisme model context words se target word ko predict karta hai.
# Example: "I love ___ learning" mein model "machine" ko predict karega.

# 1. CBOW (Continuous Bag of Words) Architecture
# CBOW mein network ek funnel (chaudai se patla) ki tarah kaam karta hai. Multiple inputs andar aate hain aur ek single output bahar nikalta hai.

# Input Layer: Yahan aapke context words aate hain. Agar aapka window=2 hai, toh 4 words (2 aage ke, 2 peeche ke) input banenge. 
# Ye saare words aapke neural network mein One-Hot Encoded (OHE) vectors ke form mein enter karte hain.

# Hidden Layer (The Magic Matrix): Ye sabse zaroori layer hai. 
# Is layer mein utne hi neurons hote hain jitna aap apna vector_size define karte hain (jaise aapne pichle code mein vector_size=5 rakha tha).

# Specialty: Baaki neural networks ki tarah isme koi activation function (jaise ReLU ya Sigmoid) nahi hota.
# Ye bas saare input context vectors ko weight matrix ke sath multiply karke unka Average nikal leta hai.

# Output Layer: Ye ek Softmax classifier hota hai. Iska size aapki total vocabulary (saare unique words) ke barabar hota hai. 
# Ye har word ko ek probability score (0 se 1 ke beech) deta hai. Jiska score sabse high hoga, model kahega: "Mujhe lagta hai blank space mein ye target word aana chahiye."


### NOTE :  for small data use CBOW and for large data use Skip-gram

## Skip-gram
# Skip-gram architecture mein model target word se context words ko predict karta hai.
# Example: "machine" se model "I", "love", "learning" ko predict karega.

# 2. Skip-gram Architecture
# Skip-gram bilkul CBOW ka ulta (reverse) hai. Ye ek patle pipe se shuru hokar aage ja kar fail (expand) jata hai.

# Input Layer: Yahan sirf ek single word aata hai (Aapka Target word, jaise "machine"),
#  wo bhi One-Hot Encoded vector ke form mein.

# Hidden Layer: Ye layer CBOW ki tarah hi same kaam karti hai. Single input vector weights ke sath multiply hota hai. 
# Iska size bhi aapke vector_size (e.g., 5, 100, ya 300) ke barabar hota hai aur isme bhi koi activation function nahi hota.

# Output Layer: Yahan ek Softmax layer nahi hoti, balki multiple Softmax classifiers hote hain (utne hi jitne aapke context words predict karne hain).
# Agar window=2 hai, toh model char alag-alag probability distributions nikalega: "Is word ke 2 step left, 1 step left, 1 step right, aur 2 step right mein kaunse words aane chahiye?"

```

```python
import gensim 
import os
```

```python
import nltk 
from nltk import sent_tokenize
from gensim.utils import simple_preprocess

nltk.download('punkt')  #machine learning model hai jo sentence todne ka basic math aur pattern samajhta hai.
nltk.download('punkt_tab') #Ye English language ke punctuation, grammar aur sentence todne ke actual rules ka ek heavy dataset hai (lagbhag 10-20 MB ka). Engine ko chalne ke liye ye fuel chahiye hota hai

story = []
for filename in os.listdir('data'):
    # f = open(os.path.join('data', filename), encoding='utf-8', errors='ignore')  # to remove UnicodeDecodeError used encoding
    with open(os.path.join('data', filename), encoding='utf-8', errors='ignore') as f:   # isme with use kiya h jo automatically f.close kr deta h 
        corpus = f.read()
        raw_sent = sent_tokenize(corpus)
        for sent in raw_sent:
            story.append(simple_preprocess(sent))

len(story)
```

```python
story[3]
```

```python
# now model build up 
model = gensim.models.Word2Vec(window=10,min_count=2)

model.build_vocab(story)
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
# import plotly.express as px 
# fig = px.scatter_3d(X[:100], x=0, y=1, z=2, color=y[:100])
# fig.show()

import pandas as pd
import plotly.express as px

# 1. Total words mein se pehle 100 words nikaliye (Kyunki aap X[:100] plot kar rahe hain)
words = model.wv.index_to_key[:100]

# 2. X, Y, Z aur unke Words ko ek Table (DataFrame) mein daal dijiye
df = pd.DataFrame({
    'Word': words,
    'X': X[:100, 0],
    'Y': X[:100, 1],
    'Z': X[:100, 2]
})

# 3. Ab Plotly se boliye ki wo is DataFrame ko plot kare aur text bhi dikhaye
fig = px.scatter_3d(
    df, 
    x='X', 
    y='Y', 
    z='Z',
    color='Word',  # Har word ko alag color dega
    text='Word',   # Dots ke upar actual shabd likha aayega
    title="Word2Vec 3D Visualization (Top 100 Words)"
)

# Thodi styling taaki words clearly padhne mein aayein
fig.update_traces(textposition='top center', marker=dict(size=5))
fig.show()
```
