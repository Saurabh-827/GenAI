```python
# OHE( One Hot Encoding )
# OHE is a technique used to convert categorical data into numerical data.
# In OHE, each unique word in the corpus is represented as a binary vector.
# For example, if we have a corpus with the following sentences:
# Sentence 1: "I love machine learning"
# Sentence 2: "Machine learning is great"
# Corpus for the above sentences would be: I love machine learning machine learning is great
# The vocabulary for this corpus would be: ["I", "love", "machine", "learning", "is", "great"]
# The OHE representation for the above sentences would be:
# Sentence 1:  
# [
#   [1, 0, 0, 0, 0, 0], // I   
#   [0, 1, 0, 0, 0, 0], // love
#   [0, 0, 1, 0, 0, 0], // machine
#   [0, 0, 0, 1, 0, 0]  // learning
# ]
# Sentence 2: 
# [
#   [0, 0, 1, 0, 0, 0], // machine
#   [0, 0, 0, 1, 0, 0], // learning
#   [0, 0, 0, 0, 1, 0], // is
#   [0, 0, 0, 0, 0, 1]  // great
# ]

# pros
# - Simple and easy to understand
# - Works well for small vocabularies
# cons
# - Sparsity- Sparsity is the presence of a large number of zero values in the feature vectors. foe example - [1, 0, 0, 0, 0, 0]
# - no fixed size - The size of the feature vectors can vary depending on the size of the vocabulary. For example, if we have a corpus with 1000 unique words, the feature vectors will have 1000 dimensions.
# - Out of vocabulary - If a new word is encountered that is not in the vocabulary, it cannot be represented in the feature vector. For example, if we have a corpus with the vocabulary ["I", "love", "machine", "learning", "is", "great"] and we encounter the word "awesome", it cannot be represented in the feature vector.
# - No capturing of sementic - OHE does not capture the semantic meaning of words. For example, the words "love" and "like" would be represented as different vectors, even though they have similar meanings.
```

```python
## Bag Of Words(BOW) 
# BOW is a technique used to convert text data into numerical data.
# In BOW, each unique word in the corpus is represented as a count of the number of times it appears in a document.
# For example, if we have a corpus with the following sentences:

# Sentence 1: "I love machine learning"
# Sentence 2: "Deep learning is complex from machine learning"

# The corpus (all combined text) for the above sentences would be: 
# "i love machine learning deep learning is complex from machine learning"

# The vocabulary (list of unique words) for this corpus would be (Total 8 words): 
# ["i", "love", "machine", "learning", "deep", "is", "complex", "from"]

# The BOW representation for the above sentences would be:
# Sentence 1: [1, 1, 1, 1, 0, 0, 0, 0] 
# (Counts for: i=1, love=1, machine=1, learning=1, deep=0, is=0, complex=0, from=0)

# Sentence 2: [0, 0, 1, 2, 1, 1, 1, 1] 
# (Counts for: i=0, love=0, machine=1, learning=2, deep=1, is=1, complex=1, from=1)

```

```python
import pandas as pd
df  = pd.DataFrame({'text':['I love machine learning','Deep learning is complex from machine learning']})


from sklearn.feature_extraction.text import CountVectorizer
## here single words are not counted 
cv = CountVectorizer()

bow = cv.fit_transform(df['text'])


print(cv.vocabulary_)
print(bow.toarray())
```

```python
# Advsntsges of BOW:
# - handle different size by using same vocabulary 
# - handle out of vocubalry 
# - captures semantic to some extent
# Disadvantages of BOW:
# - sparsity:
# - ignoring out of vocabulary 
# - ignoring word order
# - think same positive and negative sentences 
```

```python
## N- Grams 
# N-grams are contiguous sequences of n items from a given sample of text or speech.
# For example, if we have a sentence "I love machine learning", the 2-grams (bigrams) would be:
# ["I love", "love machine", "machine learning"]
# The 3-grams (trigrams) would be:
# ["I love machine", "love machine learning"]
# The N-gram representation for the above sentence would be:
# 2-grams: [1, 1, 1]
# 3-grams: [1, 1]
# Advantages of N-grams:
# - captures word order to some extent
# - captures semantic to some extent
# Disadvantages of N-grams:
# - sparsity
# - ignoring out of vocabulary
# - slows down 
```

```python
## TF-IDF
# TF-IDF stands for Term Frequency-Inverse Document Frequency.
# TF-IDF is a technique used to convert text data into numerical data.
# TF-IDF is calculated as follows:
# TF-IDF = TF * IDF
# where:
# TF (Term Frequency) is the number of times a term appears in a document in compare to total number of terms in the document. 
# For example, if we have a document with 100 words and the term "machine" appears 5 times, the TF for "machine" would be 5/100 = 0.05.
# IDF (Inverse Document Frequency) is the logarithm of the total number of documents divided by the number of documents containing the term. 
# For Example, if we have a corpus with 1000 documents and the term "machine" appears in 100 documents, the IDF for "machine" would be log(1000/100) = log(10) = 1.
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
tfidf.fit_transform(df['text']).toarray()
```

```python
print(tfidf.idf_)      # idf values increased by 1 in sklearn because of smoothing to avoid division by zero error when a term is not present in any document.
print(tfidf.get_feature_names_out())
```

```python
# tfidf used in infromation retrieval like search engine

# Advantages of tf-idf
# - captures word importance to some extent
# - captures semantic to some extent
# Disadvantages of tf-idf
# - sparsity
# - out of vocabulary
# - big dimensionality 
# - semantic
```
