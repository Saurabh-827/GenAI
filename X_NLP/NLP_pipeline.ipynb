```python

## What is NLP pipeline? 
# NLP pipeline are the sets of steps followed to build an end to end NLP software.
# Steps of NLP pipeline
  # 1. Data Acquisition(collection) - Collecting the data from various sources like web, social media, etc.
  # 2. Text Preparation-
        # a. Text Cleanup
        # b. Basic Preprocessing
        # c. Advance Preprocessing 
  # 3. Feature Engineering
  # 4. Modeling -
        # a. Model Building
        # b. Evaluation
  # 5. Deployment-
        # a. Deploy
        # b. Monitoring
        # c. Model Update 

## Points to remember 
# 1. NLP pipeline is not a linear process, it is an iterative process.
# 2. This Pipeline is not universal. 
# 3. Deep learning pipeline is slightly different.
```

```python
## Data Acquisition 
# Different scenarios while collecting data 
# Available : 1- We have data and we know the source of data.
            # 2- Data in database - I have to communicate with the data engineering team .
            # 3- Get less data - Generate more data by using any of these ways - synonym replacement, bigram flip, back translation, add noise etc.          
# External Sorces :  We don't have data but we know the source of data. For Example -  public dataset, web scrapping, image , pdf , audio , API etc 
# Not available :  We don't have data and we don't know the source of data. 
                   # for example -  I have to talk with the product team to get some data from the users and use manual labeling on that data.
```

```python
## Text Preparation

# Cleanup - Perform different cleaning steps like - removing html tags, removing emojis , check spell mistakes etc.
# Basic Preprocessing - basic steps - tokenization(sentence or word tokens),  
                       #optional preprocessing-  stop word removal, lemmatization, stemming, removing digits etc.
# Advance Preprocessing- Part Of Speech tagging (can be if not done stop word removal), parsing, coreference resolution etc.
```

```python
## Feature Engineering (Text Vectorization)
# Feature Engineering is the process of transforming the raw data into input data.
# Different types of features -
# 1. Bag of Words (BoW) - It is a simple representation of text data where each word is represented as a vector of its frequency in the document.
# 2. Term Frequency-Inverse Document Frequency (TF-IDF) - It is a representation of text data where each word is represented as a vector of its frequency in the document and the inverse document frequency.
# 3. Word Embeddings - It is a representation of text data where each word is represented as a vector of its meaning.
# 4. N-grams - It is a representation of text data where each word is represented as a vector of its frequency in the document and the frequency of its neighboring words.
# 5. Word2Vec - It is a representation of text data where each word is represented as a vector of its meaning and the meaning of its neighboring words.
# 6. One Hot Encoding - It is a representation of text data where each word is represented as a vector of its presence or absence in the document.


## Differences occurs in feature engineering stage while working with Machine learning and deep learning pipeline.
# In machine learning pipeline, we use traditional feature engineering techniques like BoW, TF-IDF, etc. and we manually generate features according to our domain  knowledge. (interpretable)
# In deep learning pipeline, we use word embeddings like Word2Vec, GloVe, etc. and we let the model learn the features from the data. we uses neural network in this . (uninterpretable)
```

```python
## Modeling
# In this stage we train our model on the prepared data and evaluate the model performance.
# In this stage we use two steps : 
# 1. Model Building : We uses different approaches to model our system depending of amount of data and nature of problem: 
                    # Heuristic Approach , machine learning approach , deep learning approach , cloud api.
# 2. Evaluation : Evaluation done by two methods intrinsic evaluationa and extrinsic evaluation.
# Intrinsic evaluation - It is a method of evaluating the performance of a model by comparing the output of the model with a gold standard dataset. 
# (Jab aap khud ke data pe test krte ho) (technical level)
# Extrinsic evaluation - It is a method of evaluating the performance of a model by comparing the output of the model with the output of a human. 
# (jab ye business me as a product users ke through us ekiya jaata h) (business level)
```

```python
## Deployment

# In this stage we deploy our model in production and monitor the performance of the model.
# Deployment - It is the process of making the model available for use in production.
# Monitoring - It is the process of monitoring the performance of the model in production.
# Model Update - It is the process of updating the model in production with new data and new features.
```
