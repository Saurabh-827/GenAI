```python
## creating tuples
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
## We can convert list to tuple and tuple to list 

# List to tuple
tup = tuple([1,2,3])
print(tup)

# Tuple to list
lst = list((1,2,3));
print(lst)
```

```python
# we can have different type element in tuple like a list 
mixed_tuple = (1, "Hello", 3.14, True, [1, 2, 3])
print(mixed_tuple)
```

```python
## Accessing tuple elements 
my_tuple = (1,2,'word')
my_list = list(my_tuple)

print(my_tuple[0])
print(my_tuple[2])
print(my_tuple[-1])

print(my_tuple[0:2])
print(my_tuple[1:2])
print(my_list[1:2])
```

```python
# tuple operation
# concatenation

tuple1 = (1, 2, 3)
tuple2 = (4, 'hello', 6)
concatenated_tuple = tuple1 + tuple2
print(concatenated_tuple)
```

```python
multiply_tuple = (1,1, 2, 3, 'helo')
print(multiply_tuple * 3)

## Tuples are immutable in nature 

# multiply_tuple[0] += 1    'tuple' object does not support item assignment 
```

```python
## Tuples methods 

print(multiply_tuple.count(1))  #it will count how many times 1 present in multiply_tuple
print(multiply_tuple.index('helo'))  # it will print index of helo in tuple
```

```python
## Packing and unpacking tuple

packed_tuple  = 1, 2, 'hello'

print(packed_tuple)


# unpacked_tuple

a, b, d = packed_tuple
print(a, b , d)
```

```python
## Unpacking with *

numbers  = (1, 2, 3, 4, 5, 6)

firs,  *middle,  last = numbers 
print(firs)
print(middle)  # starred variable always return list
print(last)
```

```python
## nested tuple 

nested_list = [[1,2,3,4], [5,6], ['heleo', 'why']]

nested_tuple = ((1, 2, 3), (4, 5, 6), ('a', 'b', 'c'))

##accesing nested tuples 

print(nested_list[0][3])
print(nested_tuple[2][1])

#slicing operation on tuple or list 

print(nested_tuple[0][0:2])
```

```python
#iterating over nested tuple 

for nest in nested_tuple:
    for item in nest:
        print(item)
```
