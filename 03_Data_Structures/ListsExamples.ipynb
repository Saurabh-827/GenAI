#### Real-World Examples Using Lists in Python

##### Example 1. Manage A To Do List
- Create a To Do List To Keep Track OF Tasks

```python
to_do_list = ["Buy Groceries","Clean House", "Pay Bills" ]
print(to_do_list)

#Adding task 

to_do_list.append("Attend meeting")
to_do_list.append("Go for run")

print(to_do_list)

#removing task

to_do_list.remove("Clean House")
print(to_do_list)

#Checking if a task is in the list 

if "Pay Bills" in to_do_list:
    print("you have to pay bills")

print("To do list remaining")
for task in to_do_list:
    print(f'[]{task}')
```

##### Example 2: Organizing Student Grades
- Create a list to store and calculate average grades for students

```python
grades = [85, 94, 78, 89, 49]

grades.append(98)
print(grades)

#Calculating average grades

average_grades = sum(grades) // len(grades)

print(f'Average grades: {average_grades}')


#finding highest and lowest grade

highest_grade = max(grades)
lowest_grade = min(grades)

print(f'{highest_grade} is the highest grade')
print(f'{lowest_grade} is the lowest grade')
```


##### Example 3: Managing An Inventory
- Use a list to manage inventory items in a store

```python
inventory = [
    "Laptop",
    "Mouse",
    "Keyboard",
    "Monitor"
]

inventory.append("Printer")
print(inventory)

inventory.remove("Keyboard")
print(inventory)

#checking if item in stock 
item = "Mouse"
if item in inventory:
    print(f"{item} is in stock")
else:
    print(f"{item} is not in stock")

#Printing the inventory

print("Current Inventory:")
for item in inventory:
    print(f'- {item}')
```

##### Example 4:Collecting User Feedback
- Use a list to collect and analyze user feedback.

```python
feedback = [
    "Great product!",
    "Could be better",
    "Amazing experience",
    "Needs improvement"]

feedback.append("Will buy again")
print(feedback)

feedback.remove("Could be better")
print(feedback)

#checking if feedback contains a specific comment

positive_feedback_count = sum(1 for comment in feedback if "great" in comment.lower() or "amazing" in comment.lower())
print(f'Positive Feedback Count: {positive_feedback_count}')

#printing feedback 
for comment in feedback:
    print(f'-{comment}')
```
