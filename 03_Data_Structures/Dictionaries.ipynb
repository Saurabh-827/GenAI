```python
## Creating Dictionaries
# most used
empty_dict = {}   # {} in this we have to write keys in quotes like {"name": "Saurabh", "age": 24}
print(type(empty_dict))
```

```python
empty_dict = dict() # dict() in this we can write keys without quotes like dict(name= "sau", age= 24 )
print(type(empty_dict))

# thoda slow — pehle function call hoti hai

# # {} se numbers as keys ho sakte hain
# d = {1: "one", 2: "two"}  # ✅

# # dict() se numbers as keys nahi ho sakte
# d = dict(1="one")  # ❌ SyntaxError

```

```python
person = {"name": "Saurabh", "age": 24}
print(person)

person2 = dict(name= "sau", age= 24 )
print(person2)
```

```python
# In case of duplicate keys, the last value is always used
person = {"name": "Saurabh", "age": 24, "job": "No"}
person = {"name": "Saurabh", "age": 24, "job": "No", "job": "yes"}

print(person)
```

```python
# Accessing Dictionary Elements
person = {"name": "Saurabh", "age": "24", "job": "No"}

print(person["name"])
print(person["age"])
# print(person["skills"]) # this will raise a KeyError

# Accessing using built-in get() method

print(person.get("name"))  #in real projects get mostly used 
print(person.get('age'))
print(person.get("skills"))  #this will return none 
print(person.get("skills", "Key not found"))  #this will return the default value

#                     []	            get()
# Key exist hai	    ✅ value milegi	✅ value milegi
# Key exist nahi	❌ KeyError	    ✅ None milega
# Default value	    ❌	            ✅ de sakte ho
```

```python
# Modifying Dictionary Elements 
# Keys cannot be modified but values can be added, updated or deleted

person = {"name": "Saurabh", "age": "24", "job": "No"}
person["age"] = 25  #update
print(person)

person["skills"] = ("Python", "Java")  #add
print(person)

del person['age']  #delete
print(person)
```

```python
## Dictionary Methods 

keys = person.keys() #return all the keys 
print(keys)

values = person.values() # return all the values 
print(values)

items = person.items()  # return all the key-value pairs 
print(items)
```

```python
# Shallow Copy  (important)

person_copy = person
print(person_copy)

person["name"] = "Sau"
print(person_copy)  # this will also change the name in person_copy because both person and person_copy are referencing the same dictionary in memory   
```

```python
person = {"name": "Saurabh", "age": "24", "job": "No"}

perosn_copy1 = person.copy() #copy creates a new dictionary with the same key-value pairs
person["job"] = "yes"
print(person)
print(perosn_copy1)
```

```python
# Iterating over Dictionaries 
# we can iterate over dictionaries, keys , values, or items 

# Iterating over keys 
for key in person.keys():
    print(key)

#iterating over values
for value in person.values():
    print(value)

#iterating over items 
for key, value in person.items():
    print(f'{key}: {value}')
```

```python
#Nested Dictionaries

persons = {
    "person1": {"name": "saurabh", "age" : 24},
    "person2": {"name": "sau", "age" : 25}
}
print(persons)
```

```python
#access nested dictionaries 

print(persons["person1"])
print(persons["person1"]["name"])
```

```python
#iterating over nested dictionaries

for person_id, person_info in persons.items():
    print(f'{person_id} : {person_info}')
    for key, value in person_info.items():
        print(f'{key} : {value}')

```

```python
##Dictionary comprehension

squares = {x: x**2 for x in range(5)}
print(squares)

## Conditional dictionary conprehension

even_squares = { x: x**2 for x in range(20) if x%2==0}
print(even_squares)
```

```python
# Practical Examples 

# Use dictionary to count number frequency of every elements in list

numbers = [1, 2, 3, 4, 5, 1, 2, 3, 1, 2, 1]

frequency ={}

for number in numbers: 
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1

print(frequency)
```

```python
## Merging Dictionaries into one 

dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "d": 4}

# method 1 using update() method
merged_dict = dict1.copy()  # create a copy of dict1 to avoid modifying it
merged_dict.update(dict2)  # update the merged_dict with dict2
print(merged_dict)

# method 2 using dictionary unpacking (Python 3.5+)
merged_dict = {**dict1, **dict2}  # unpack both dictionaries into a new dictionary
print(merged_dict)

# method 3 using the union operator (Python 3.9+)
merged_dict = dict1 | dict2  # merge using the union operator
print(merged_dict)


```
