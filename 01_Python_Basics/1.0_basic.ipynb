Syntax is the set of rules for correctly structering statements.

```python
## Basic syntax rules in Python
# 1. Case Sensitivity
name="Saurabh"
Name="Harshita"

print(name)
print(Name)
```

```python
# 2. Indentation
## Python uses indentation to define blocks of code.
age=10
if age > 18:
    print("You are an adult.")

print("You are not an adult.")
```

```python
#3. Line Continuation
num= 1+2+3+4+5+\
3+1
print(num)
```

```python
## multiple statements in single line
a=10; b=20; c=a+b
print(c)
```

```python
## Understand semantics in python
# variable assignment 
name="SH"
type(name)
```

```python
#Type Inference
age=25
print(type(age))

age="Twenty-Five"
type(age)
```
