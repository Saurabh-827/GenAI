```python
## Data Types are the classification that specifies which type of value a variable has and
# what kind of mathematical, relational, and logical operations can be applied to it.

#Common  basic datatypes in python
#1. Integer: Represents whole numbers without a fractional part.
x = 10 
print(type(x))

#2. Float: Represents numbers with a fractional part (decimal point).
y = 3.14
print(type(y))

#3. String: Represents a sequence of characters enclosed in quotes.
name="Saurabh"
print(type(name))

#4. Boolean: Represents a value that can be either True or False.
is_true = True
print(type(is_true))
```
