```python
## Declaring and assigning variables 
name="Saurabh"
age=24
print("My name is", name)
print("My age is", age)
```

```python
## Naming Conventions
## variables name should be descriptive
## It must start with letters or with an underscore(_).
# 
# Invalid Naming like starting with a number, symbols and using different operators between it.

my_name="Saurabh"
print(my_name)

# my@name = "Saurabh"   #invalid
```

```python
## Understanding Different Data Types in Python
## Python is dynamically typed lanugauage it means datatypes are defined during run time of the code 

is_rich = True  #boolean type
print(type(is_rich))
```

```python
## Type checking and conversion

print(type(name))
print(type(age))
print(type(str(age)))
```

```python
## Dynamic Typing
# Python allows type of the variables to change as program execute 
var = 10
print(type(var) , type(str(var)))
```

```python
## Input and type casting

user_input=input("What is your name?")
print("Hello", user_input)
print("Your name has data type of", type(user_input))

age_input = int(input("What is your age?"))  #TypeCasting

print("Your age is", age_input)
print("Your age has data type of", type(age_input))
```

```python
# Simple Calculator

num1 = int(input("Enter First Number"))
num2 = int(input("Enter Second Number"))

sum = num1 + num2
difference = num1 - num2
product = num1 * num2
quotient = num1 / num2

print("Sum:", sum)
print("Difference:", difference)
print("Product:", product)
print("Quotient:", quotient)
```
