```python
## Map Function
# The map function applies a given function to each item of an iterable (like a list) and returns a map object (which is an iterator). 
# You can convert this map object into a list or other iterable types.

def square(n):
    return n**2

numbers = [1, 2, 3, 4, 5, 6]
print(list(map(square, numbers)))

#using lambda function

print(list(map(lambda m: m**2, numbers)))
```

```python
# using map() with multiple iterables 

num1 = [1,2,3]
num2 = [4,5,6]

added_nums = list(map(lambda x,y: x+y, num1, num2))
print(added_nums)
```

```python
## using map() to convert a list of strings to integers

str_numbers = ['1', '2', '3', '4', '5']

numbers = map(int, str_numbers)

print(list(numbers))
```

```python
words = [
    'hello', 'world', 'python', 'programming'
]

upper_words = map(str.upper, words)

print(list(upper_words))
```

```python
def get_name(person):
    return person['name']

people = [
    {'name': 'Harshita', 'age': 24},
    {'name': 'Harsh', 'age': 16},
    {'name': 'Sau', 'age': 23}
]

names = map(get_name, people)
print(list(names))
```
