```python
def even_or_odd(num):
    """This function checks even or odd"""
    if(num%2==0):
        return "Even"
    else:
        return "Odd"

even_or_odd(10)
```

```python
# function with multiple parameters 

def add_nums(a,b):
    return a+b

add_nums(10,20)
```

```python
## default parameters

def greet(name="guest"):
    return "Hello "+name

greet("Sau")
```

```python
## Variable length argument 
# Positional and keywords arguments


# Positional arguments are values passed to a function based on the order in which the function's parameters were defined.
def print_num(*arguments):   # *args is used to accept multiple positional arguments 
    for num in arguments:
        print(num)


print_num(10, 20, 30, 40)
```

```python
# Keyword Arguments --  keyword arguments allow you to pass values to a function by explicitly stating the parameter name (e.g., name="Alice").
# This differs from positional arguments, which rely on the order in which you provide them

def print_detail(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_detail(name="Babua", age=30, city="New York")
```

```python
# passing positional parameters and keyword parameters together

def print_info(*args, **kwargs):
    for arg in args:
        print(arg)
    for k,v in kwargs.items():
        print(f"{k}: {v}")

print_info("Hello", "World", name="Babua", age=30)
```
