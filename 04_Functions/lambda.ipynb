```python
## Lambda Functions
# Lambda functions are anonymous functions that can be defined in a single line. 
# They are often used for short, simple functions that are not reused elsewhere in the code.

#syntex
lambda arguments: expression
```

```python
addition = lambda a,b : a + b
print(type(addition))
print(addition(5,3))
```

```python
even = lambda num:num%2 == 0

print(even(456))
```

```python
addmore = lambda a,b,c,d: a+b+c+d
print(addmore(1,2,3,4))
```

```python
## using map()

numbers = [1,2,3,4,5]

list(map(lambda num:num**2, numbers))
```
